# 🔍 Universal Approximation Theorem in Action: Solving XOR

This notebook demonstrates the **Universal Approximation Theorem** using the classic XOR classification problem — a perfect example where a linear model fails but a tiny neural network succeeds.

---

## 🧠 What is the Universal Approximation Theorem?

> A feedforward neural network with a **single hidden layer** and a **non-linear activation function** can approximate any continuous function on a compact domain, given enough neurons.

In simpler terms:  
Neural networks don’t need to be deep to be powerful — they just need **non-linearity** and a bit of width.

Let’s prove it by solving the XOR problem.

---

## 🧩 Step 1: Define the XOR Dataset


```python
import numpy as np

# Inputs and outputs for XOR
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])
```

---

## ❌ Step 2: Try a Linear Model (Logistic Regression)


```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='liblinear', max_iter=1000)
model.fit(X, y)
print("Predictions:", model.predict(X))
```
